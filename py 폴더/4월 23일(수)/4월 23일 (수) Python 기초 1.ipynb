#변수란 컴퓨터의 기억장소에  이름을 부여한 것을 말한다
#컴퓨터의 기억공간에 이름을 부여하고 그 공간에 값을 저장하거나 읽어올 수 있다
#Markdown은 전체를 설명으로 바꾼다
#컴퓨터는 0과 1로 구성되어 있다
#JAVA라면 변수를 미리 선언해야 한다. int a;
#int 형의 테이터가 저장될 공간을 확보하고 이름을 a라고 해라
#보통의 인터프리터 언어는 변수를 미리 선언하지 않는다. 필요할 때 만들어서 쓰면 된다. 


```python
a = 10 #a라는 변수에 값 10을 저장(할당)하라 
b = 20 #b라는 변수에 값 20을 저장하라 
c = a+b #a의 값 10과 b의 값 20을 읽어와서 더한 뒤에 c에 넣어라
        #좌변에는 변수만, 우변에는 아무거나 올 수 있다
print(a,b,c) #a,b,c 변수의 값을 읽어서 화면에 출력하라

```

```python
#값 입력하기 - 입력함수가 동시에 출력도 가능

a = input("a =")
b = input("b =")
c = a+b #모든 입력은 unicode다 그래서 a도 b도 문자로 인식한다 

print(a,b,c) #문자 + 문자 = 문자 덧붙히기

d = int(a) + int(b) #int는 문자를 순자로 바꾸는 역할을 한다

print(a,b,d)
```

```python
#사각형의 면적 구하기 (가로, 세로 입력받아서 면적을 구하고자 한다)
#입력데이터 : 가로, 세로
#출력데이터 : 면적
#입력 -> 계산이나 출력을 생각하지 말자 (중요, 한번에 한번씩 처리해야 한다)

width = input("가로 : ")
height = input("세로 : ")
#계산
area = int(width) * int(height)
#출력
print(area)

```

```python
#주급 계산기: 시간당 급여액하고 근무 시간을 입력 받아서 주급을 계산하기

hours = input("근무 시간 : ")
wage = input("시간 급여액 : ")

weekly_wage = int(hours) * int(wage)

print("근무시간은", hours, "시간당급여액",wage, "주급은", weekly_wage)
```
